## <font color='red'> **Questions : 가장 많은 Tip을 받으려면 어떤 요일, 시간대? 어떤 유형의 손님에게 하여야 하나?**


돈을 더 받기 위해서 평균적으로 팁을 더 많이 주는 일요일이 좋습니다.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# tips 데이터셋 다시 로드 (이전 세션에서 정의되지 않았을 수 있으므로 재정의)
tips = sns.load_dataset('tips')

# 스타일 설정 및 시각화
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# 요일별 평균 팁 시각화 (일요일을 더 돋보이게 하기 위해 일요일만 붉은 톤으로 강조)
colors = ['lightgray' if day != 'Sun' else '#ff6b6b' for day in ['Thur', 'Fri', 'Sat', 'Sun']]

sns.barplot(
    data=tips,
    x='day',
    y='tip',
    order=['Thur', 'Fri', 'Sat', 'Sun'],
    errorbar=None,
    palette=colors,
    hue='day',
    legend=False
)

plt.title('Average Tip Amount by Day (Sunday is the Best!)', fontsize=14, fontweight='bold')
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Tip ($)', fontsize=12)
plt.ylim(0, 4.0)

# 값 표시
for p in plt.gca().patches:
    height = p.get_height()
    if height > 0:
        plt.gca().annotate(f'${height:.2f}',
                            (p.get_x() + p.get_width() / 2., height + 0.1),
                            ha='center', va='center',
                            fontsize=11, color='black', fontweight='bold' if height > 3.2 else 'normal')

plt.tight_layout()
plt.show()
```

## 데이터분석은 연구자의 의사결정과 관련이 있다.
- 어떤 시간대 유형의 손님을 찾는 것은 연구자에 따라 다르다.

지도를 그려서 median_house_value/population 별로 scatter 하여 어느지역의 집값이 비싼지 말해줘? 시각화도 해줘.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 파일 로드
file_path = '/content/sample_data/california_housing_train.csv'
df_housing = pd.read_csv(file_path)

# 데이터 정보 확인
display(df_housing.head())
print(df_housing.info())
```

```python
plt.figure(figsize=(10, 8))

# 경도(longitude)를 x축, 위도(latitude)를 y축으로 하여 캘리포니아 실제 지형 모양의 산점도 시각화
# 주택 가격 중앙값(median_house_value)에 따라 색상(cmap='jet') 변화
# 인구수(population)에 따라 방울의 크기(s) 조절
scatter = plt.scatter(
    x=df_housing['longitude'],
    y=df_housing['latitude'],
    alpha=0.4,
    s=df_housing['population'] / 100,  # 가독성을 위해 인구수 비례 스케일 조절
    c=df_housing['median_house_value'],
    cmap=plt.get_cmap('jet'),
    label='Population'
)

plt.colorbar(scatter, label='Median House Value')
plt.title('California Housing Prices Map (Value vs Population)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend()
plt.show()
```

### 캘리포니아 집값 분석 요약

시각화 결과를 보면 다음과 같은 특징을 파악할 수 있습니다:
1. **해안가 지역(Coastal Areas)**: 경도와 위도 기준으로 태평양 해안가를 따라 붉은색(높은 주택 가격) 점들이 밀집되어 있습니다.
2. **주요 대도시권**: 특히 북부의 **샌프란시스코 베이 지역(San Francisco Bay Area)**과 남부의 **로스앤젤레스(Los Angeles)** 및 **샌디에이고(San Diego)** 인근 지역의 집값이 압도적으로 높고 인구 밀도(원의 크기)도 매우 높게 나타납니다.
3. **내륙 지역(Inland)**: 내륙으로 갈수록 주택 가격 중앙값이 낮아지는 경향(파란색 점들)을 뚜렷하게 관찰할 수 있습니다.

### 팁을 가장 많이 받을 수 있는 요일 분석

기존에 로드된 `tips` 데이터를 바탕으로 요일별 평균 팁, 팁 총합, 그리고 방문 고객 수를 분석하여 최적의 요일을 찾아보겠습니다.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# tips 데이터셋 다시 로드 (이전 세션에서 정의되지 않았을 수 있으므로 재정의)
tips = sns.load_dataset('tips')

# 스타일 설정 및 시각화
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# 요일별 평균 팁 시각화 (일요일을 더 돋보이게 하기 위해 일요일만 붉은 톤으로 강조)
colors = ['lightgray' if day != 'Sun' else '#ff6b6b' for day in ['Thur', 'Fri', 'Sat', 'Sun']]

sns.barplot(
    data=tips,
    x='day',
    y='tip',
    order=['Thur', 'Fri', 'Sat', 'Sun'],
    errorbar=None,
    palette=colors,
    hue='day',
    legend=False
)

plt.title('Average Tip Amount by Day (Sunday is the Best!)', fontsize=14, fontweight='bold')
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Tip ($)', fontsize=12)
plt.ylim(0, 4.0)

# 값 표시
for p in plt.gca().patches:
    height = p.get_height()
    if height > 0:
        plt.gca().annotate(f'${height:.2f}',
                            (p.get_x() + p.get_width() / 2., height + 0.1),
                            ha='center', va='center',
                            fontsize=11, color='black', fontweight='bold' if height > 3.2 else 'normal')

plt.tight_layout()
plt.show()
```

### 💡 분석 결과 및 전략 제안

1. **건당 평균 팁 액수가 가장 높은 요일: 일요일 (Sunday)**
   * 일요일의 평균 팁은 **$3.26**으로 요일 중 가장 높습니다. (중앙값 또한 $3.15로 가장 높음)
   * 테이블당 팁을 두둑하게 챙겨주는 알짜배기 손님들은 일요일에 많이 방문합니다.

2. **전체 팁 총액 및 일거리가 가장 많은 요일: 토요일 (Saturday)**
   * 토요일은 방문 테이블 수가 **87회**로 압도적으로 많으며, 이에 따라 얻을 수 있는 전체 팁 합계도 **$260.40**으로 가장 큽니다.
   * 박리다매 형식으로 끊임없이 서빙하며 누적 팁을 최대로 쌓고 싶다면 토요일 근무가 유리합니다.
